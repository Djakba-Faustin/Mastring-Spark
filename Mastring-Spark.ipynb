## What is Spark Architecture?

Apache Spark follows a **master-slave architecture** where the Driver acts as the master and Executors are the slaves. Tasks are distributed across worker nodes to enable parallel processing.
### Core Architectural Components
* **Driver Program**: The central process that runs the `main()` function of your application. It creates the **SparkContext** (or `SparkSession`), which coordinates and schedules jobs.
* **Cluster Manager**: An external resource allocation service (e.g., YARN, Mesos, Kubernetes, Standalone) that coordinates resources between the Driver and the Workers.
* **Worker Node**: Any machine in the cluster capable of running application code.
* **Executor**: A dedicated process launched on a Worker Node for an application. It runs individual **Tasks** and stores data in its **Cache**.
* **Task**: The smallest individual unit of execution sent by the Driver to an Executor to process a specific block of data.
## Spark Execution Flow
1. **Submit Job** via driver program.
2. **Driver creates DAG** of stages.
3. **Tasks** are sent to executors.
4. **Executors** perform computation and return results to driver.
## DAG: Directed Acyclic Graph
* **Represents** the logical execution plan.
* **Spark breaks job into stages** (based on wide/narrow transformations).
* **Within each stage**, it executes **tasks** in parallel.
### Example: DAG in Action


```python
from pyspark.sql import SparkSession
from pyspark.sql.types import StructType, StructField, StringType, IntegerType
import pyspark.sql.functions as F

# Initialize Spark Session cleanly
spark = SparkSession.builder \
    .appName("DAG Example") \
    .getOrCreate()

# 1. Define explicit schema (Best practice for performance & data integrity)
schema = StructType([
    StructField("name", StringType(), True),
    StructField("amount", IntegerType(), True)
])

# 2. Define Sample Data
data = [("Alice", 100), ("Bob", 200), ("Alice", 300)]

# 3. Create DataFrame (Triggers no Spark Jobs - purely builds the initial DAG)
df = spark.createDataFrame(data, schema=schema)

# 4. Add a Transformation (Triggers a 'Wide' transformation / Shuffle)
# This appends a grouping stage to the execution plan without executing it yet.
grouped_df = df.groupBy("name").agg(F.sum("amount").alias("total_amount"))

# 5. Trigger an Action (This executes the complete DAG plan)
grouped_df.show()
```

### Explanation of the Execution:

* `groupBy().sum()` creates a **logical plan** but does not execute immediately.
* When `.show()` is called, Spark triggers the job and constructs a **DAG** (Directed Acyclic Graph).
* Spark identifies **narrow and wide transformations** and breaks the job into **stages** and **tasks** accordingly.



### 🔥 Lazy Evaluation

Spark delays execution until an **action** is called (like `.show()`, `.collect()`). This allows it to:

* **Optimize the execution plan** (via the Catalyst Optimizer).
* **Avoid unnecessary computations** by knowing the final target upfront.

# Module 4: Working with RDDs

## What is an RDD?

* **RDD (Resilient Distributed Dataset)** is the core abstraction of Spark.
* An **immutable**, distributed collection of objects that can be processed in parallel.
* Each dataset in an RDD is divided into **partitions**.
* Operations are **lazy**, fault-tolerant, and memory-efficient.


### 🔐 Key Features

| Feature | Description |
| :--- | :--- |
| **Immutable** | Once created, it cannot be changed. |
| **Fault-Tolerant** | Uses lineage info to recompute lost data. |
| **Partitioned** | Data is split across nodes. |
| **Lazy Evaluation** | Transformations aren't executed until an action is called. |


## RDD Operations

### ✅ Transformations (lazy)

* **Return a new RDD** instead of modifying the existing one.
* **Not executed immediately** until an action is triggered.

**Examples:**
`map()`, `filter()`, `flatMap()`, `union()`, `distinct()`, `groupByKey()`, `reduceByKey()`


### 🚀 Actions (trigger execution)

* **Return values** to the driver program or **write data** to external storage systems.
* **Triggers the actual computation** of the built DAG.

**Examples:**
`collect()`, `count()`, `take()`, `reduce()`, `foreach()`, `saveAsTextFile()`

### 📝 Example: Create RDD and Perform Transformations

The code below demonstrates how to:
1. Create a distributed RDD from a local Python list.
2. Apply a lazy **transformation** (`map`) across the partitions.
3. Call an **action** (`collect`) to pull the processed data back to the driver.



```python
# Initialize Spark (in Databricks this is already done)
rdd = spark.sparkContext.parallelize([1, 2, 3, 4, 5])

# Transformation: Multiply each element by 2
rdd2 = rdd.map(lambda x: x * 2)

# Action: Collect results
print(rdd2.collect())
```

## 🧠 When to Use RDDs vs DataFrames

| Use Case | RDDs | DataFrames |
| :--- | :---: | :---: |
| **Low-level transformations** | ✅ | ❌ |
| **Compile-time type safety (Scala/Java)** | ✅ | ❌ |
| **Ease of use and performance** | ❌ | ✅ |
| **Optimized query engine** | ❌ | ✅ (Catalyst) |
## 🔄 RDD Persistence (Caching)

If an RDD will be reused, you can **cache** or **persist** it to speed things up.
This keeps it in memory across operations (unless memory is insufficient).


```python
# Caching / Persisting an RDD to memory for faster reuse
rdd2.persist()  # or rdd2.cache()
```

## 🧪 Exercise: Filter Even Numbers from a List

This exercise uses an RDD `.filter()` transformation with a lambda function to extract only the even values from a distributed collection.


```python
# Define local list and parallelize into an RDD
nums = [10, 15, 20, 25, 30]
rdd = spark.sparkContext.parallelize(nums)

# Filter out elements that aren't divisible by 2
evens = rdd.filter(lambda x: x % 2 == 0)

# Pull the final filtered results back to the driver
print(evens.collect())
```

# Industry Track: Building Reliable Spark Applications

The fundamentals above explain how Spark works. This section adds the practices expected in professional data platforms: explicit schemas, deterministic transformations, data-quality checks, performance awareness, observability, testing, security, and deployable application design.

## Production Principles

1. **Make data contracts explicit.** Define schemas, required fields, valid ranges, and ownership.
2. **Prefer built-in Spark functions.** They are usually faster and easier for Spark to optimize than Python UDFs.
3. **Keep transformations deterministic and testable.** Separate input, transformation, and output responsibilities.
4. **Measure before tuning.** Use `explain()`, Spark UI metrics, row counts, and runtime measurements.
5. **Design for reruns.** Use stable keys, checkpoints, idempotent writes, and controlled output modes.
6. **Protect data.** Apply least privilege, secret management, encryption, masking, and audit logging.

## Professional Learning Outcomes

```python
from datetime import date
from decimal import Decimal

from pyspark.sql import functions as F
from pyspark.sql.window import Window
from pyspark.sql.types import (
    DateType,
    DecimalType,
    IntegerType,
    StringType,
    StructField,
    StructType,
)

sales_schema = StructType([
    StructField("order_id", StringType(), False),
    StructField("customer_id", StringType(), False),
    StructField("order_date", DateType(), False),
    StructField("category", StringType(), False),
    StructField("quantity", IntegerType(), False),
    StructField("unit_price", DecimalType(10, 2), False),
])

sales_data = [
    ("O100", "C01", date(2026, 1, 3), "Electronics", 2, Decimal("199.99")),
    ("O101", "C02", date(2026, 1, 4), "Books", 3, Decimal("24.50")),
    ("O102", "C01", date(2026, 1, 8), "Books", 1, Decimal("39.00")),
    ("O103", "C03", date(2026, 1, 11), "Electronics", 1, Decimal("899.00")),
]

sales = (
    spark.createDataFrame(sales_data, sales_schema)
    .withColumn("revenue", F.col("quantity") * F.col("unit_price"))
)

customers = spark.createDataFrame(
    [("C01", "North"), ("C02", "South"), ("C03", "North")],
    ["customer_id", "region"],
)

customer_sales = sales.join(customers, on="customer_id", how="inner")
window_by_customer = Window.partitionBy("customer_id").orderBy(F.col("order_date").desc())

customer_sales.select(
    "order_id", "customer_id", "region", "category", "revenue",
    F.row_number().over(window_by_customer).alias("latest_order_rank"),
).show()

customer_sales.groupBy("region", "category").agg(
    F.sum("revenue").alias("total_revenue"),
    F.countDistinct("customer_id").alias("unique_customers"),
).orderBy("region", "category").show()
```

## Data Quality and Data Contracts

A production job should reject or quarantine invalid records instead of silently producing incorrect metrics. Typical checks include schema validity, uniqueness, nullability, referential integrity, accepted values, and numerical ranges.

```python
def validate_sales_quality(frame):
    metrics = {
        "row_count": frame.count(),
        "null_order_ids": frame.filter(F.col("order_id").isNull()).count(),
        "duplicate_order_ids": (
            frame.groupBy("order_id").count()
            .filter(F.col("count") > 1)
            .count()
        ),
        "invalid_quantities": frame.filter(F.col("quantity") <= 0).count(),
        "invalid_prices": frame.filter(F.col("unit_price") < 0).count(),
    }
    violations = {
        name: value
        for name, value in metrics.items()
        if name != "row_count" and value
    }
    if violations:
        raise ValueError(f"Sales data quality checks failed: {violations}")
    return metrics

quality_report = validate_sales_quality(sales)
print("Quality checks passed:", quality_report)

# Built-in expressions keep logic inside Spark's optimizer.
clean_sales = (
    sales.withColumn("category", F.initcap(F.trim("category")))
    .withColumn("revenue", F.round("revenue", 2))
    .filter(F.col("quantity") > 0)
)
clean_sales.explain(mode="formatted")
```

## Performance Engineering

Spark performance depends on data size, partitioning, serialization, shuffles, and file layout. Production tuning should be evidence-based:

- Filter and select only the columns needed as early as possible.
- Avoid unnecessary `collect()` calls because they move data to the driver.
- Use `repartition()` when increasing parallelism and `coalesce()` when reducing partitions.
- Cache only data that is reused and materialize the cache deliberately.
- Broadcast a small dimension table when it avoids an expensive shuffle.
- Prefer Parquet or Delta in production for column pruning and predicate pushdown.
- Inspect the physical plan and Spark UI before changing configuration.

```python
# Compare a normal join with an explicitly broadcast dimension join.
# Broadcast only when the dimension is genuinely small enough for executor memory.
small_customers = F.broadcast(customers)
optimized_sales = sales.join(small_customers, on="customer_id", how="inner")
optimized_sales.select("order_id", "region", "revenue").show()

print("Optimized join plan:")
optimized_sales.explain(mode="formatted")
```

## Structured Streaming

Structured Streaming uses the DataFrame API for unbounded data. A streaming job should define its input schema, transformations, checkpoint location, output mode, trigger policy, and recovery behavior. The local example below builds a stream without leaving a query running.

```python
stream_schema = StructType([
    StructField("event_id", StringType(), False),
    StructField("event_type", StringType(), False),
    StructField("event_time", DateType(), False),
])

# The rate source is useful for local demonstrations. Replace it with Kafka,
# Event Hubs, files, or another supported source in a real application.
streaming_events = (
    spark.readStream.format("rate")
    .option("rowsPerSecond", 5)
    .load()
    .select(
        F.col("value").cast("string").alias("event_id"),
        F.lit("heartbeat").alias("event_type"),
        F.to_date("timestamp").alias("event_time"),
    )
)

streaming_summary = (
    streaming_events.groupBy("event_type")
    .count()
)
print("Streaming DataFrame created:", streaming_summary.isStreaming)
print("Output mode should be chosen deliberately: append, update, or complete.")
```

## Machine Learning with MLlib

A Spark ML pipeline makes feature preparation and model training reproducible. The pipeline below uses a categorical feature, a numeric feature, a train/test split, and an evaluation metric. In production, persist the fitted pipeline and version the training data, features, and model configuration.

```python
from pyspark.ml import Pipeline
from pyspark.ml.classification import LogisticRegression
from pyspark.ml.evaluation import BinaryClassificationEvaluator
from pyspark.ml.feature import OneHotEncoder, StringIndexer, VectorAssembler

model_data = spark.createDataFrame(
    [
        ("North", 120.0, 1),
        ("North", 80.0, 0),
        ("South", 45.0, 0),
        ("South", 160.0, 1),
        ("North", 210.0, 1),
        ("South", 60.0, 0),
        ("North", 30.0, 0),
        ("South", 190.0, 1),
    ],
    ["region", "revenue", "label"],
)

region_indexer = StringIndexer(inputCol="region", outputCol="region_index", handleInvalid="keep")
region_encoder = OneHotEncoder(inputCol="region_index", outputCol="region_vector")
assembler = VectorAssembler(inputCols=["region_vector", "revenue"], outputCol="features")
classifier = LogisticRegression(featuresCol="features", labelCol="label", maxIter=20)

pipeline = Pipeline(stages=[region_indexer, region_encoder, assembler, classifier])
train_data, test_data = model_data.randomSplit([0.75, 0.25], seed=42)
model = pipeline.fit(train_data)
predictions = model.transform(test_data)

predictions.select("region", "revenue", "label", "prediction", "probability").show(truncate=False)
auc = BinaryClassificationEvaluator(labelCol="label", metricName="areaUnderROC").evaluate(predictions)
print(f"Test ROC AUC: {auc:.3f}")
```
